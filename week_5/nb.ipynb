```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sine Wave")
plt.plot(x, y)

plt.title("Sine Wave")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
%pip install seaborn

import seaborn as sns
import pandas as pd

# Create a simple DataFrame
data = {"Category": ["A", "B", "C", "D", "E"], "Value": [23, 45, 56, 12, 39]}
df = pd.DataFrame(data)

# Create a bar plot
sns.barplot(x="Category", y="Value", data=df)
plt.title("Category Values")
plt.show()
```
